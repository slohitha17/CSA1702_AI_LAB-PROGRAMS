```python
colors=['R','G','B']
graph={'A':['B','C'],'B':['A','C','D'],'C':['A','B','D'],'D':['B','C']}
color={}

def solve(v):
    if v==len(graph): return True
    x=list(graph)[v]
    for c in colors:
        if all(color.get(n)!=c for n in graph[x]):
            color[x]=c
            if solve(v+1): return True
            del color[x]
    return False

solve(0)
print(color)
```
