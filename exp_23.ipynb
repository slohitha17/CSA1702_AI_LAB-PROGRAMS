```python
def ab(depth,alpha,beta,maximizing):
    if depth==0: return 3
    if maximizing:
        v=-999
        for _ in range(2):
            v=max(v,ab(depth-1,alpha,beta,False))
            alpha=max(alpha,v)
            if alpha>=beta: break
        return v
    else:
        v=999
        for _ in range(2):
            v=min(v,ab(depth-1,alpha,beta,True))
            beta=min(beta,v)
            if alpha>=beta: break
        return v

print("Best value:",ab(3,-999,999,True))
```
