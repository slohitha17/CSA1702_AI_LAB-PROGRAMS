```python
female = ["pam", "liz", "ann"]

parent = {
    "pam": ["bob"],
    "tom": ["bob"],
    "bob": ["jim"]
}

mother = input("Enter mother: ").lower()
child = input("Enter child: ").lower()

if mother in female and child in parent.get(mother, []):
    print(mother, "is mother of", child)
else:
    print("Relation not found")
```
