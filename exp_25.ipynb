```python
from sklearn.neural_network import MLPClassifier

X=[[0,0],[0,1],[1,0],[1,1]]
y=[0,1,1,0]

m=MLPClassifier(hidden_layer_sizes=(4,),max_iter=2000,
                random_state=1).fit(X,y)

x=list(map(int,input("Enter two values: ").split()))
print("Output:",m.predict([x])[0])
```
