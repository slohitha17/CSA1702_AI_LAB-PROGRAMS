```python
def minimax(depth,maximizing):
    if depth==0: return 3
    if maximizing:
        return max(minimax(depth-1,False),minimax(depth-1,False))
    return min(minimax(depth-1,True),minimax(depth-1,True))

print("Best value:",minimax(3,True))
```
