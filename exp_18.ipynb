```python
from itertools import permutations

n=int(input("Cities: "))
cost=[list(map(int,input().split())) for _ in range(n)]
best=10**9

for p in permutations(range(1,n)):
    path=(0,)+p
    d=sum(cost[path[i]][path[i+1]] for i in range(n-1))+cost[path[-1]][0]
    if d<best: best=d; route=path

print("Route:",route+(0,))
print("Cost:",best)
```
