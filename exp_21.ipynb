```python
b=[' ']*9

def show():
    print(b[0:3]); print(b[3:6]); print(b[6:9])

for i in range(9):
    p=int(input("Position 1-9: "))-1
    b[p]='X' if i%2==0 else 'O'
    show()
    for a,c,d in [(0,1,2),(3,4,5),(6,7,8),(0,3,6),(1,4,7),(2,5,8),(0,4,8),(2,4,6)]:
        if b[a]!=' ' and b[a]==b[c]==b[d]:
            print(b[a],"Wins"); exit()
```
