```python
def bird(x):
    return x == "parrot"

def has_wings(x):
    return bird(x)

def can_fly(x):
    return has_wings(x)

x = input("Enter bird: ").lower()

if can_fly(x):
    print(x, "can fly")
else:
    print(x, "cannot fly")
```
