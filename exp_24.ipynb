```python
from sklearn.tree import DecisionTreeClassifier

X=[[1],[2],[3],[4]]
y=[0,0,1,1]

m=DecisionTreeClassifier().fit(X,y)
x=int(input("Enter value: "))
print("Prediction:",m.predict([[x]])[0])
```
