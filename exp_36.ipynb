```python
graph = {
    "A": [("B", 2), ("C", 1)],
    "B": [("D", 5)],
    "C": [("D", 2)],
    "D": []
}

path = ["A"]
current = "A"

while current != "D":
    next_node = min(graph[current], key=lambda x: x[1])[0]
    path.append(next_node)
    current = next_node

print("Path =", path)
```
