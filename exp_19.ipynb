```python
import heapq

g={'A':{'B':1,'C':4},'B':{'D':2},'C':{'D':1},'D':{}}
h={'A':4,'B':2,'C':1,'D':0}

q=[(h['A'],0,'A',[])]
while q:
    f,c,v,p=heapq.heappop(q)
    if v=='D':
        print("Path:",p+[v],"Cost:",c); break
    for n,w in g[v].items():
        heapq.heappush(q,(c+w+h[n],c+w,n,p+[v]))
```
