```python
g={'A':['B','C'],'B':['D','E'],'C':['F'],'D':[],'E':[],'F':[]}

def dfs(v,seen=set()):
    if v not in seen:
        print(v,end=" ")
        seen.add(v)
        for x in g[v]: dfs(x,seen)

dfs(input("Start: "))
```
