```python
from collections import deque

start=tuple(map(int,input("Enter 9 values: ").split()))
goal=(1,2,3,4,5,6,7,8,0)
q=deque([(start,[])])
moves=[(-1,0),(1,0),(0,-1),(0,1)]

while q:
    s,path=q.popleft()
    if s==goal:
        print("Solution:",path); break
    z=s.index(0); x,y=divmod(z,3)
    for dx,dy in moves:
        a,b=x+dx,y+dy
        if 0<=a<3 and 0<=b<3:
            n=a*3+b; t=list(s)
            t[z],t[n]=t[n],t[z]
            if tuple(t) not in [x[0] for x in q]:
                q.append((tuple(t),path+[tuple(t)]))
```
