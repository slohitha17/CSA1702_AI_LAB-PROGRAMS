```python
r,c = map(int,input("Rows Columns: ").split())
A=[list(map(int,input().split())) for _ in range(r)]
B=[list(map(int,input().split())) for _ in range(r)]

C=[[A[i][j]+B[i][j] for j in range(c)] for i in range(r)]
for row in C: print(row)
```
