```python
from collections import deque

g={'A':['B','C'],'B':['D','E'],'C':['F'],'D':[],'E':[],'F':[]}
start=input("Start: ")
q=deque([start]); seen=set()

while q:
    v=q.popleft()
    if v not in seen:
        print(v,end=" ")
        seen.add(v)
        q.extend(g[v])
```
