```python
from collections import deque

start=(3,3,0); goal=(0,0,1)
q=deque([(start,[start])]); seen=set()

def valid(s):
    m,c,b=s
    return 0<=m<=3 and 0<=c<=3 and (m==0 or m>=c) and (3-m==0 or 3-m>=3-c)

while q:
    s,p=q.popleft()
    if s==goal:
        print(*p,sep="\n"); break
    if s in seen: continue
    seen.add(s)
    m,c,b=s
    for x,y in [(1,0),(2,0),(0,1),(0,2),(1,1)]:
        n=(m-x,c-y,1) if b==0 else (m+x,c+y,0)
        if valid(n): q.append((n,p+[n]))
```
